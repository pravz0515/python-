```python
#list comprehension
#syntax:[exp for iterator in iterable if condition]
#return 1 - n numbers as list using List comprehension
n=int(input())
numbers=[i for i in range(1, n + 1)]
print(numbers)
```

```python
#first n even numbers
n=int(input())
num=[x for x in range(0,2*n,1) if x%2==0]
print(num)
```

```python
# return list even or odd, for list of integers
arr = [20,34,33,55,23,54]
res=["odd" if x&1 else "Even" for x in arr ]
print(res)
```

```python
#even - add 1
#odd- add 0.5
arr = [20, 34, 33, 55, 23, 54]
res = [x + 0.5 if x & 1 else x + 1 for x in arr]
print(res)
```

```python
#find one die combinations 
print([x for x in range(1,7)])
```

```python
#find two die combinations 
print([(i,j) for i in range(1,7) for j in range(1,7)])
```

```python
#find two die combinations if any dice face is 4
print([(i,j) for i in range (1,7) for j in range(1,7) if i==4 or j==4])
```

```python
#find the number is positive even,negative even ,positive odd or negative odd
n=int(input())
res=[("positive even" if n%2==0 else "positive odd") if n>-1 else ("negative even" if n%2==0 else "negative odd")]
print(res)
```
