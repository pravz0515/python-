```python
#Types of Arguments
#2.Keyword Arguments:the arguments are assinged to parametersas per positional order
def sub(x:int, y:int)->int:
    c=x - y
    return c
#function calling block
a , b =10, 20
print(sub(x = a, y = b))
print(sub(y = b, x = a))
```

```python
#Types of Arguments
#2.Keyword Arguments:the arguments are passed along with the parameters
def sub(x:int, y:int)->int:
    c=x - y
    return c
#function calling block
a , b =10, 20
print(sub(x = a, y = b))
print(sub(y = b, x = a))
```

```python
def sub(x:int, y:int, z:int, w:int)->int:
    p=x + y
    q=x+y+z
    r=x+y+w
    return p , q , r
#function calling block
a , b , c , d=10, 20, 30, 40
print(sub(x = a, y = b, z = c, w = d))
print(sub(y = b, x = a, w = d , z = c))
```

```python
#Types of Arguments
#3.Keyword Arguments:the arguments are assinged to parametersas per positional order
def add(x:int, y:int)->int:
    c=x + y
    return c
#function calling block
a , b =10, 20
res = add(a,b)
print(res)
```

```python
#Default Arguments:The parameters defined with some default values,and it assign the argument value if you pass the values for default
# parameters otherwise, it will assign default values
def add(a,b,c = 0 , d = 0):
    res = a + b + c + d
    return res
print(add(10,20))
print(add(10,20,5))
print(add(10,20,2,1))
```

```python
 #Default Arguments:
def mul(a,b,c = 1 , d = 1):
    res = a*b*c*d
    return res
print(mul(10,20))
print(mul(10,20,5))
print(mul(10,20,2,1))
```

```python
# 4. variable length of positional arguments
def add(a, b = 0 , *args):
    print(type(args))
    print(args)
    s = 0
    for num in args:
        s += num
    return s
print(add(1))
#print(add(1, a=2, b=3))
print(add(1,2,3,4,5,6,7,8,9))
```

```python
# 4. variable length of positional arguments
def add(a, b = 0 , *args):
    print(type(args))
    print(args)
    s = 0
    for num in args:
        s += num
    return s + a + b
print(add(1))
#print(add(1, a=2, b=3))
print(add(1,2,3,4,5,6,7,8,9))
```
