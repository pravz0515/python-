```python
import re
```

```python
#match(pattern,string)
string = "The sky is blue and Ocean is blue"
res=re.match("[A-Z]",string)
print(res)
print(res.span()) #it returns the matched positions
print(res.group())#it returs matched pattern val
```

```python
#match(pattern,string)
string = "The sky is blue and Ocean is blue"
res=re.match("[A-Za-z]+",string)
print(res)
print(res.span()) #it returns the matched positions
print(res.group())#it returs matched pattern val
```

```python
#match(pattern,string)
string = "The sky is blue and Ocean is blue"
res=re.match("z",string)
print(res)
print(res.span()) #it returns the matched positions
print(res.group())#it returs matched pattern val
```

```python
#match(pattern,string)
try:
    string = "The sky is blue and Ocean is blue"
    res=re.match("z",string)
    print(res)
    print(res.span()) #it returns the matched positions
    print(res.group())#it returs matched pattern val
except Exception as e:
    print("there is no matched pattern:",e)
```

```python
#match(pattern,string)
#search 5 length word
try:
    string = "The sky is blue and Ocean is blue"
    res=re.search("[A-Za-z]{5}",string)
    print(res)
    
except Exception as e:
    print("there is no matched pattern:",e)

```

```python
#get the words from string which length is morethan 3
#findall(pattern,string)
try:
    string = "The sky is blue and Ocean is blue"
    res=re.findall("[A-Za-z]{3,}",string)
    print(res)
    
except Exception as e:
    print("there is no matched pattern:",e)
```

```python
try:
    string = "The sky is blue and Oc11ean1223 is blue"
    res=re.findall("[A-Za-z]{1,}",string)
    print(res)
    
except Exception as e:
    print("there is no matched pattern:",e)
```

```python
# validate mobile number
#all are digits,length of number is 10,starts with 6,7,8,9,Remaining are 0-9
try:
    mobile = input()
    res=re.findall("^[6-9][0-9]{9}",mobile)
    print(res)
    if res:
        print(f"{mobile} is valid")
    else:
        print(f"{mobile} is invalid")
    
except Exception as e:
    print("there is no matched pattern:",e)
```

```python
# validate pancard number
# alphanumeric,length of number is 10,starts with A-Z,Remaining are 0-9
import re
try:
    mobile = input()
    res=re.fullmatch(r"^[A-Z]{5}[0-9]{4}[A-Z]$",mobile)
    print(res)
    if res:
        print(f"{mobile} is valid")
    else:
        print(f"{mobile} is invalid")
    
except Exception as e:
    print("there is no matched pattern:",e)
```
