```python
import math

print(math.ceil(4.956))      # Upper bound value
print(math.floor(4.75))      # Lower bound value
print(math.factorial(5))      # Factorial
print(math.gcd(10, 5, 2))     # Greatest Common Divisor
print(math.lcm(45, 5))        # Least Common Multiple
print(math.log10(2))          # Log base 10
print(math.log10(4))
print(math.log2(4))           # Log base 2
print(math.pi)                # Pi value
```

```python
import random

print(random.random())          # Random float between 0 and 1

print(random.randint(1, 10))    # Random integer between 1 and 10

print(random.choice([
    'A', 1, 2, 3, 4, 5, 6, 7, 8, 9,
    'J', 'Q', 'K'
]))                              # Returns one random element

print(random.choices(
    ['A', 1, 2, 3, 4, 5, 6, 7, 8, 9,
     'J', 'Q', 'K'],
    k=3
))

arr = ['A', 1, 2, 3, 4, 5, 6, 7, 8, 9,
       'J', 'Q', 'K']

random.shuffle(arr)

print(arr)
```

```python
import random

steps = 0
rolls = 0

while True:
    input("To roll dice click Enter button ")

    dice_val = random.randint(1, 6)
    rolls += 1

    if steps + dice_val <= 100:
        steps += dice_val

    if steps == 100:
        print(f"Your reached TOP in {rolls} rolls")
        break

    print(f"For {rolls} rolls your at {steps} position")
```
