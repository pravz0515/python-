```python
#create mobile class
class Mobile:
    company = 'Iphone' #class attribute
    def __init__(self, model, ram , rom, price):
        self.model = model     #instance attributes
        self.ram = ram         #instance attributes
        self.rom = rom         #instance attributes
        self.price = price     #instance attributes`
#object creation
iphone16 = Mobile("iphone 16",16,256,125000)
iphone17 = Mobile("iphone 17",8,256,125000)
print(iphone16.price)
print(iphone17.price)
print(iphone16.company)
print(iphone17.company)
print(Mobile.company)

```
