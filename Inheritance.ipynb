```python
#types of inheritance
#1.Single inheritance
#with implementation in child class
#when child dont have constructor in this case parent class constructor behaves like child class constructor
#parent class
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def details(self):
        return f"My name is {self.name} and I am a {self.age} years old"
#child class
class Student(Person):
    pass
s = Student("prava",22)
print(s.name)
print(s.age)
print(s.details())
```

```python
#parent class
class Person:
    def __init__(self,name,age,gender):
        self.name=name
        self.age=age
        self.gender= gender
    def details(self):
        return f"My name is {self.name} and I am a {self.age} years old {self.gender}"
#child class
class Student(Person):
    def __init__(self,stdid,name,age):
        self.stdid = stdid
        self.name = name
        self.age = age
p = Person("pandu",22,"female")
s = Student(101,"prava",22)
print(s.name)
print(s.age)
s.gender="female" #adding 
print(s.details())
print(p.details())
```

```python
# Inherete the parent class constructor in child class by using super()
#super():This function is user to get the parent class implementation in child class
#parent class
class Person:
    def __init__(self,name,age,gender):
        self.name=name
        self.age=age
        self.gender= gender
    def details(self):
        return f"My name is {self.name} and I am a {self.age} years old {self.gender}"
#child class
class Student(Person):
    def __init__(self,stdid,name,age,gender):
        self.stdid = stdid
        super().__init__(name,age,gender)
    def percentage(self,marks:list):
        return sum(marks)/len(marks)
s = Student(101,"prava",22,"female")
print(s.name)
print(s.age)
print(s.details())
print(s.percentage([90,89,88]))
```

```python
#multi level inheritance
class employee:
    Company="cognizant"
    def __init__(self,eid,name,salary):
        self.eid=eid
        self.name=name
        self.salary=salary
    def calculate_salary(self,bonus=0):
        return self.salary+bonus
class teamlead(employee):
    def __init__(self,eid,name,salary,project_name):
        self.project_name = project_name
        super().__init__(eid,name,salary)
    def project_status(self,status):
        return f"{self.project_name} project status: {status}% completed"
class manager(teamlead):
    def __init__(self,eid,name,salary,project_name,no_projects):
        self.no_projects=no_projects
        super().__init__(eid,name,salary,project_name)
e=employee(101,"pravalika",70000)
t=teamlead(102,"pravs",850000,"healthecare")
m = manager(103,"prava",60000,"aiml",3)
print(e.name,t.name,m.name)
print(e.calculate_salary())
print(t.calculate_salary(10000))
print(t.project_status(90))
print(m.calculate_salary(10000))
print(m.project_status(90))
```

```python
#multiple inheritance
class father:
    def __init__(self,father_name):
        self.father_name=father_name
    def working(self):
        return "father working as business"
    
class mother:
    def __init__(self,mother_name):
        self.mother_name=mother_name
    def working(self):
        return "mother is a house wife"
class child(father,mother):
    def __init__(self,name,father_name,mother_name):
        self.name = name
        mother.__init__(self,mother_name)
        father.__init__(self,father_name)
    def working(self):
        f = father(self.father_name)
        m = mother(self.mother_name)
        print(f.working())
        print(m.working())
c = child("nenu","ma nanna","ma amma")
print(c.father_name)
print(c.mother_name)
print(child.__mro__) # mro represents method resolution order
print(c.working())
   
```
