```python
#sorted(iterable,reverse=False,Key=function):It retuerns the new sorted iterable 
#sort the fruits based on the length of fruit name
fruits=['Apple','banana','kivi','Mango','Water Melon']
res = sorted(fruits)
print(res)
print(sorted(fruits,reverse=True))
print(sorted(fruits,reverse=True,key=len)) # it sorts the list based on the length of each item in list
```

```python
#map(function, iterable)
#find the square of all list of integers using map function
def square(x) :
    return x**2
arr = [1,2,3,4]
res = list(map(square, arr))
print(res)
```

```python
#filter(function,iterbale)
#find even numbers from list of numbers
def even(x):
    if x%2==0:
        return x
lst=[1,2,3,4,5]
res=list(filter(even,lst))
print(res)
```

```python
def even(x):
    if x%2==0:
        return "even"
lst=[1,2,3,4,5]
res=list(filter(even,lst))
print(res)
```

```python
#find the 3's multipliers from list of integers
def multipliers(x):
    if x%3==0:
        return x
arr = [1,24,54,22,44,51,33,15]
res=list(filter(multipliers,arr))
print(res)


```

```python
#enumerate(iterable)
fruits=['Mango','Apple','banana','kivi','Water Melon']
print(enumerate(fruits))
print(list(enumerate(fruits)))
for ind, val in enumerate(fruits) :
    print(f"{ind}-{val} and length is {len(val)}")
    
```

```python
percentage = [34,35,54,65,67]
for num in percentage:
    if num<35 :
        print("All students not passed")
    break
else:
    print("All students passed")
```

```python
percentage = [34,35,54,65,67]
lst = []
for num in percentage:
    if num < 35 :
        lst.append(False)
else:
        lst.append(True)
print(lst)
if all(lst):
    print("All Students Paas")
else:
    print("All students not passed")
```
