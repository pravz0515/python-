```python
from abc import ABC, abstractmethod
class mobile(ABC):
    @abstractmethod
    def Ram(self):
        pass
    @abstractmethod
    def Rom(self):
        pass
    @abstractmethod
    def Battery(self):
        pass
    #concrete method
    def about_device(self):
        return "This is mobile "
class oppo(mobile):
    def __init__(self,model,o_ram,o_rom,o_battery):
        self.model=model
        self.o_ram=o_ram
        self.o_rom=o_rom
        self.o_battery=o_battery
    def Ram(self):
        return f"The oppo {self.model} have {self.o_ram}GB RAM"
    def Rom(self):
        return f"The oppo {self.model} have {self.o_rom}GB ROM"
    def Battery(self):
        return f"The oppo {self.model} have {self.o_battery} mah"
    def camera(self,front,back):
        return f"The oppo {self.model} have {front}MP front camera + {back}MP back camera"
#m = mobile() #cannot instantiate for abstract class
reno16 = oppo("reno16",16,256,5000)
reno15 = oppo("reno15",18,256,10000)
print(reno16.about_device())
print(reno16.Ram())
print(reno15.Ram())

```
