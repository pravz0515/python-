```python
#access values byusing keys
details={"name":"pravalika","age":22,'course':'python','marks':[1,2,3],'age':200}
print(details['course'])
print(details['age'])
#updating dictionary value
details['age'] +=1
details['course']='java'# updates value for course key
print(details)
# item assignment
details['location']='hyd'
print(details)
```

```python
#dictionary methods
details={"name":"pravalika","age":22,'course':'python','marks':[1,2,3],'age':50}
print(details.get('name'))
print(details.get('location'))
print(details.get('age',18))
print(details.get('location','Location key not exist'))
```

```python
details={"name":"pravalika","age":22,'course':'python'}
d={'marks':[1,2,3],'age':50}
details.update(d)
print(details)
```

```python
details={"name":"pravalika","age":22,'course':'python','marks':[1,2,3],'age':50}
print(details.pop('age'))
print(details.popitem())
print(details)
```

```python
details={"name":"pravalika","age":22,'course':'python','marks':[1,2,3],'age':50}
print(details.keys())
print(details.values())
print(details.items())
```

```python
details={"name":"pravalika","age":22,'course':'python','marks':[1,2,3],'age':50}
keys=details.keys()
print(type(keys))
print(keys)
print(details.values())
print(details.items)
print(list(keys))
```
