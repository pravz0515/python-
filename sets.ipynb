```python
s={1,2,3,4,5,6,67,7}
s1={1,2,3,8}
print(s.union(s1))
print(s.intersection(s1))
print(s1.difference(s))
print(s.symmetric_difference(s1))
```

```python
s={1,2,3,4,5,6,67,7}
s1={1,2,3,8}
s2={3,5,6,7,8}
print(s.union(s1,s2))
print(s.intersection(s1,s2))
print(s & s1)
print(s1.difference(s,s2))
print(s.symmetric_difference(s1,s2))
```

```python
s={1,2,3,4,5,6,67,7}
s1={1,2,3,8}
s2={3,5,6,7,8}
print(s | s2) #union
print(s & s2) #intersection
print(s - s1) #difference
print(s1 ^ s2) #symmetric_difference
```
