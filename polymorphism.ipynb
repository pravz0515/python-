```python
#Method Overriding
class person:
    def __init__(self,name,age):
        self.name= name
        self.age= age
    def details(self):
        return  f"{self.name} {self.age}"
class student(person):
    def __init__(self,name,age,rollno):
        self.rollno= rollno
        super().__init__(name,age)
    def details(self):
        return  f"my name is {self.name} and i am a {self.age} with {self.rollno}"
s = student("pravalika",22,7)
print(s.details())
p = person("prav",18)
print(p.details())
```

```python
#method overloading
class maths:
    def add(self,x,y):
       return x+y
    def add(self,x,y,z):
        return x+y+z
    def add(self,a,b,c,d):
        return a+b+c+d
m = maths()
print(m.add(1,2))
print(m.add(1,2,3))
print(m.add(1,2,3,5))
#we can achieve this by using default arguments or orbitary arguments
```

```python
#method overloading
class maths:
    def add(self,a,b,c=0,d=0):
        return a+b+c+d
m = maths()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
#we can achieve this by using default arguments or orbitary arguments
```
