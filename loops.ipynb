```python
arr=list(map(int,input().split()))
target = int(input())
res=[]
for num in arr:
    if num != target :
        res.append(num)
print(res)
         
```

```python
arr = list(map(int, input().split()))
target = int(input())
res = []
i = 0 #initialization
while i < len(arr): #condition
    if arr[i] != target :
        res.append(arr[i])
    i += 1

print(res)
```

```python
arr = list(map(int, input().split()))
target = int(input())
i = 0
while i < len(arr):
    if arr[i] == target:
        arr.pop(i)
    else:
        i += 1

print(arr)
```
