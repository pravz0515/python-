```python
# exception handling blocks
#without exception
try:
    a = int(input())
    b = int(input())
    c = a+b
    print(c)
except:
    print("Something wrong in try block")
else:
    print("Their is no exception in try block")
finally:
    print("program done")
```

```python
# exception handling blocks
# exception with exception class
try:
    a = 10
    b = 0
    c = a/b
    print(c)
except Exception as e:
    print(f"Something wrong in try block : {e}")
```

```python
# exception handling blocks
# define multiple exceptions with different except block
try:
    a = 10
    b = 0
    c = a/b
    print(c)
except ZeroDivisionError as e:
     print(f"zero division not possible: {e}")
except IndexError as e:
     print(f"we cannot access element at this position: {e}")
except Exception as e:
    print(f"Something wrong in try block : {e}")
```

```python
# exception handling blocks
# define multiple exceptions with different except block
try:
    arr = [1,2,3,4]
    c = arr[10]
    print(c)
except ZeroDivisionError as e:
     print(f"zero division not possible: {e}")
except IndexError as e:
     print(f"we cannot access element at this position: {e}")
except Exception as e:
    print(f"Something wrong in try block : {e}")
```

```python
# exception handling blocks
# define multiple exceptions with different except block
try:
    a = 10
    c = len(a)
    print(c)
except ZeroDivisionError as e:
     print(f"zero division not possible: {e}")
except IndexError as e:
     print(f"we cannot access element at this position: {e}")
except Exception as e: #common block
    print(f"Something wrong in try block : {e}")
```
