```python
from datetime import datetime, timedelta
curr_timestamp = datetime.now() #to get current timestamp
print(curr_timestamp)
today = curr_timestamp.date()
curr_time = curr_timestamp.time()
print(today)
print(curr_time)
print(today.year)
print(today.month)
print(today.day)
print(curr_time.hour)
print(curr_time.minute)
print(curr_time.second)
print(today.strftime("%d-%m-%y"))
print(curr_timestamp.strftime("%H:%M:%S")) #to get in 24hrs format
print(curr_timestamp.strftime("%I:%M:%p")) #to get in 12hrs format
```

```python
import time
from IPython.display import clear_output
sec_counter = 0
min_counter = 0
hour_counter =0
while True:
    print(f"{hour_counter}:{min_counter}:{sec_counter}")
    time.sleep(1)
    sec_counter += 1
    if sec_counter == 60:
        sec_counter = 0
        min_counter += 1
        if min_counter == 60:
            min_counter = 0
            hour_counter += 1
            if hour_counter == 24:
                hour_counter = 0
    clear_output(wait=True)
    
    
```

```python
import time
from IPython.display import clear_output
minutes = int(input("Enter stop watch minutes:"))
sec_counter = 0
min_counter = 0
hour_counter =0
while min_counter != minutes:
    print(f"{hour_counter}:{min_counter}:{sec_counter}")
    time.sleep(1)
    sec_counter += 1
    if sec_counter == 60:
        sec_counter = 0
        min_counter += 1
        if min_counter == 60:
            min_counter = 0
            hour_counter += 1
            if hour_counter == 24:
                hour_counter = 0
    clear_output(wait=True)
    
    
```
