```python
#generator defination
def gen(x):
    x=x+1
    yield x
    x=x+2
    yield x
    x=x+3
    yield x
res=gen(5)
print(res)
print(next(res))
print(next(res))
print(next(res))
```

```python
#generator defination
#To overcome the Stop iteration error by using loops
def gen(x):
    x=x+1
    yield x
    x=x+2
    yield x
    x=x+3
    yield x
res=gen(5)
for i in res:
    print(i)
```

```python
# print the 1 to 10 numbers using function
def fun(x):
    res=[]
    for i in range(1,11):
        res.append(i)
    return res
res=fun(10)
print(res)
```

```python
# print the 1 to 10 numbers using generator
def gen(x):
    res=[]
    for i in range(1,x+1):
        yield i
res=gen(10)
for i in res:
    print(i)

        
```
